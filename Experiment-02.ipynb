```python
import sqlite3
import pandas as pd

# File path definition
csv_file_path = r"C:\Users\AI_LAB\Downloads\archive\senate_stock_discosures.csv"

# ======================================================
# STEP 1: Ingest & Inspect CSV Data
# ======================================================
print("--- Step 1: Reading from CSV ---")
df = pd.read_csv(csv_file_path)

# Display Data Summary
print("\n===== Dataset Head =====")
print(df.head())

print("\n===== Dataset Info =====")
df.info()

print("\n===== Dataset Description =====")
print(df.describe(include='all'))


# ======================================================
# STEP 2: Export Data to Various Formats
# ======================================================
print("\n--- Step 2: Exporting Data ---")

# Export to CSV
exported_csv_path = "senate_stock_disclosures_exported.csv"
df.to_csv(exported_csv_path, index=False)
print(f"Exported to CSV successfully: {exported_csv_path}")

# Export to Excel (.xlsx)
exported_excel_path = "senate_stock_disclosures_exported.xlsx"
df.to_excel(exported_excel_path, index=False)
print(f"Exported to Excel successfully: {exported_excel_path}")

# Export to SQLite Database[cite: 2]
db_name = "senate_disclosures.db"
conn = sqlite3.connect(db_name)

df.to_sql(
    name="stock_disclosures",  # Table name inside database
    con=conn,
    if_exists="replace",
    index=False
)
print(f"Exported to SQLite Database successfully: {db_name} (Table: 'stock_disclosures')")
conn.close()


# ======================================================
# STEP 3: Re-Ingest Data from Excel & SQLite DB
# ======================================================
print("\n--- Step 3: Re-ingesting Data from Excel & SQLite ---")

# Read back from Excel[cite: 2]
df_excel = pd.read_excel(exported_excel_path)
print(f"\nLoaded {len(df_excel)} rows from Excel.")
print(df_excel.head(2))

# Read back from SQLite Database[cite: 2]
conn = sqlite3.connect(db_name)
df_sql = pd.read_sql_query("SELECT * FROM stock_disclosures", conn)
print(f"\nLoaded {len(df_sql)} rows from SQLite DB.")
print(df_sql.head(2))
conn.close()
```
