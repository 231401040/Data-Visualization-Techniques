```python
import pandas as pd

# File path using raw string formatting to handle backslashes correctly
file_path = r"C:\Users\AI_LAB\Downloads\archive\senate_stock_discosures.csv"

# ======================================================
# STEP 1: Load Data & Inspect Missing Values
# ======================================================
print("--- Step 1: Detection of Missing Values ---")
df = pd.read_csv(file_path)

print(f"Total Rows Loaded: {len(df)}")
print("\nMissing values per column:")
print(df.isnull().sum())

print(f"\nTotal missing values in dataset: {df.isnull().sum().sum()}")
print(f"Any missing values present? {df.isnull().values.any()}")

# ======================================================
# STEP 2: Imputation Techniques (Filling Missing Values)
# ======================================================
print("\n--- Step 2: Applying Imputation Strategies ---")
df_imputed = df.copy()

# Separate columns by data type
numeric_cols = df_imputed.select_dtypes(include=['number']).columns
categorical_cols = df_imputed.select_dtypes(include=['object', 'category']).columns

# Impute numerical missing values using Median
for col in numeric_cols:
    if df_imputed[col].isnull().sum() > 0:
        median_val = df_imputed[col].median()
        df_imputed[col] = df_imputed[col].fillna(median_val)
        print(f"✓ Imputed numeric column '{col}' with Median: {median_val}")

# Impute categorical missing values using Mode
for col in categorical_cols:
    if df_imputed[col].isnull().sum() > 0:
        mode_val = df_imputed[col].mode()[0] if not df_imputed[col].mode().empty else "Unknown"
        df_imputed[col] = df_imputed[col].fillna(mode_val)
        print(f"✓ Imputed categorical column '{col}' with Mode: '{mode_val}'")

print(f"\nRemaining nulls after imputation: {df_imputed.isnull().sum().sum()}")

# ======================================================
# STEP 3: Alternative Approach - Dropping Missing Values
# ======================================================
print("\n--- Step 3: Demonstrating Drop Strategies ---")

# Option A: Drop any row containing missing values
df_dropped_rows = df.dropna()
print(f"Dataset shape after dropping rows with null values: {df_dropped_rows.shape}")

# Option B: Drop columns containing any missing values
df_dropped_cols = df.dropna(axis=1)
print(f"Columns remaining after dropping null columns: {df_dropped_cols.shape[1]}")

# ======================================================
# STEP 4: Save the Cleaned Dataset
# ======================================================
output_filename = r"C:\Users\AI_LAB\Downloads\archive\senate_stock_disclosures_cleaned.csv"
df_imputed.to_csv(output_filename, index=False)
print(f"\nCleaned dataset saved successfully to:\n{output_filename}")
```
