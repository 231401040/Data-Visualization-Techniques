```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Read Dataset (Updated for Colab uploaded file)
file_path = "senate_stock_discosures.csv"
df = pd.read_csv(file_path)

# 2. Use only first 10 rows
df = df.head(10)

# 3. Convert numeric columns
num = [
    "stock_price",
    "asset_value_low",
    "asset_value_high"
]
for col in num:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 4. View Data
print("First 5 Rows")
print(df.head())

print("\nShape")
print(df.shape)

print("\nData Types")
print(df.dtypes)

# 5. Filter Stock Price > 100
print("\nTransactions with Stock Price > 100")
high_price = df[df["stock_price"] > 100]
print(high_price)

# 6. Filter Purchase Transactions
print("\nPurchase Transactions")
purchases = df[df["transaction"] == "Purchase"]
print(purchases)

# 7. Sort by Stock Price
print("\nSorted by Stock Price")
sorted_df = df.sort_values(by="stock_price", ascending=False)
print(sorted_df)

# 8. Statistics on Stock Price
print("\nMean =", df["stock_price"].mean())
print("Median =", df["stock_price"].median())
print("Mode =")
print(df["stock_price"].mode())
print("Standard Deviation =", df["stock_price"].std())

# 9. Summary
print("\nSummary Statistics")
print(df.describe())

# 10. Group By Senator Last Name
print("\nAverage Stock Price by Senator")
senator_avg = df.groupby("last_name")["stock_price"].mean()
print(senator_avg)

# 11. Correlation
print("\nCorrelation")
corr = df[["stock_price", "asset_value_low"]].corr()
print(corr)

# 12. Scatter Plot
plt.scatter(df["asset_value_low"], df["stock_price"])
plt.title("Stock Price vs Asset Value (Low Range)")
plt.xlabel("Asset Value Low (USD)")
plt.ylabel("Stock Price (USD)")
plt.grid(True)
plt.show()

# 13. Bar Plot
senator_avg.plot(kind="bar")
plt.title("Average Stock Price by Senator")
plt.xlabel("Senator Last Name")
plt.ylabel("Average Stock Price (USD)")
plt.show()
```
